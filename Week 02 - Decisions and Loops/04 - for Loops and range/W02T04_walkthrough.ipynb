## 1. `for i in range(n)`

Repeats exactly `n` times. `i` takes `0, 1, 2, ...` up to but **not including** `n`.

````python
for i in range(5):
    print(i)
````

```
for i in range(5):
      i takes, in order:   0 ──▶ 1 ──▶ 2 ──▶ 3 ──▶ 4
      the indented block runs once for each value, then the loop just ends
      - no condition to check, no risk of it running forever
```

### Your turn

## 2. `range(start, stop)`

`stop` is never included. This is where most `for` mistakes happen.

````python
for i in range(1, 6):
    print(i)   # 1 2 3 4 5

for i in range(1, 5):
    print(i)   # 1 2 3 4 - not 5
````

## 3. `range(start, stop, step)`

Counting by more than one at a time — or counting down, with a negative step.

````python
for i in range(10, 0, -1):
    print(i)   # 10 9 8 ... 1
````

## 4. `for` vs `while`

Know the count in advance → `for`. Waiting on a condition or a signal → `while`
(Topic 3). `for` can loop a fixed number of records even before you have a real list of
them to loop over — Week 4 will replace `range(n)` with the real data.

````python
for i in range(3):
    value = float(input("Value: "))
    print(value)
````

## 5. `enumerate()` — the index, for free

Looping and needing to know *which one you are on* is common enough that Python
has a shortcut — `enumerate()` hands you the position and the value together, so
you never have to track a counter by hand.

````python
for i, letter in enumerate("cat"):
    print(i, letter)
````

```
enumerate("cat")  ──▶  (0, 'c'), (1, 'a'), (2, 't')
                           │         │
                      position    value — both, at the same time
```

`start=` shifts where the counting begins — useful for human-friendly numbering,
since people count from 1, not 0.

````python
for i, letter in enumerate("cat", start=1):
    print(f"letter {i}: {letter}")
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `range(5)` expecting 1–5 | gives 0–4 | starts at 0 unless you say otherwise |
| `range(1, 5)` expecting 1–5 | gives 1–4 | `stop` is never included — use `range(1, 6)` |
| using `for` when you do not know the count in advance | only works if you guess right | that is what `while` is for |
| changing `i` inside the loop, expecting it to affect the count | no effect | `range()` already decided every value in advance |
| `for i in enumerate(x):` — forgetting to unpack | `i` is a whole `(0, 'c')` pair, not just the index | unpack both: `for i, value in enumerate(x):` |
