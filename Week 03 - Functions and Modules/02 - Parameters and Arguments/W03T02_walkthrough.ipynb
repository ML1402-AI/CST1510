## 2. A default value is a fallback

Give a parameter `= something` in the `def` line, and callers can leave it out.

````python
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(95))       # uses the default, 90
print(status_of(95, 96))   # overrides it
````

## 3. Keyword arguments

Name which parameter you are filling in. Order stops mattering, and the call site
becomes readable on its own.

````python
print(status_of(percent=95, warning_at=90))
print(status_of(warning_at=90, percent=95))   # same result, reordered
````

## 4. Mixing positional and keyword

Positional arguments must come first. Once you use a keyword, everything after it
must be a keyword too.

````python
print(status_of(95, warning_at=90))    # fine
# print(status_of(percent=95, 90))    # SyntaxError
````

## 5. `*args` and `**kwargs` — when you don't know how many

Sometimes you don't know in advance how many arguments will be passed. `*args`
collects any extra positional arguments into a tuple; `**kwargs` collects extra
keyword arguments into a dict. The names `args`/`kwargs` are just convention —
the `*` and `**` are what actually matter.

````python
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))          # 6
print(total(10, 20, 30, 40))   # 100
````

```
total(1, 2, 3)
        │
        ▼
   numbers = (1, 2, 3)     - a tuple, built for you, however many you passed
```

`**kwargs` does the same job for keyword arguments, collecting them into a dict.

````python
def configure(**settings):
    for key, value in settings.items():
        print(f"{key} = {value}")

configure(host="10.0.0.1", port=8080, ssl=True)
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| a required parameter after a default one | `SyntaxError` | defaults must come last in the `def` line |
| a positional argument after a keyword one | `SyntaxError` | positional first, keyword after |
| relying on position with many parameters | wrong value in the wrong place, no error | use keywords once you have three or more |
| passing too few arguments | `TypeError: missing ... required positional argument` | check the `def` line for what it actually needs |
| forgetting `*args` arrives as a tuple, not separate variables | `numbers[0]` works, `numbers` alone is not one number | loop over it, or index into it — it's one collection |
